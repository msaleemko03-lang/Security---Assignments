```python
#التشفير (Encryption)
def encrypt_additive(plaintext, key):
    result = ""
    for char in plaintext:
        if char.isupper():
            result += chr((ord(char) - 65 + key) % 26 + 65)
        elif char.islower():
            result += chr((ord(char) - 97 + key) % 26 + 97)
        else:
            result += char
    return result
```

```python
#فك التشفير (Decryption)
def decrypt_additive(ciphertext, key):
    result = ""
    for char in ciphertext:
        if char.isupper():
            result += chr((ord(char) - 65 - key) % 26 + 65)
        elif char.islower():
            result += chr((ord(char) - 97 - key) % 26 + 97)
        else:
            result += char
    return result
```

```python
#الهجوم الأعمى (Brute Force Attack)
def brute_force_additive(ciphertext):
    for key in range(26):
        decrypted = decrypt_additive(ciphertext, key)
        print(f"Key = {key}: {decrypted}")
```

```python
#مثال تشغيل
text = "Hello World"
key = 3

cipher = encrypt_additive(text, key)
print("Encrypted:", cipher)

plain = decrypt_additive(cipher, key)
print("Decrypted:", plain)

print("\nBrute Force Attack:")
brute_force_additive(cipher)
```
