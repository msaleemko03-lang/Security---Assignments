```python
#التشفير (Encryption)
def encrypt_vigenere(plaintext, key):
    result = ""
    key = key.upper()
    key_index = 0

    for char in plaintext:
        if char.isalpha():
            k = ord(key[key_index % len(key)]) - 65

            if char.isupper():
                p = ord(char) - 65
                c = (p + k) % 26
                result += chr(c + 65)
            else:
                p = ord(char) - 97
                c = (p + k) % 26
                result += chr(c + 97)

            key_index += 1
        else:
            result += char

    return result
```

```python
#فك التشفير (Decryption)
def decrypt_vigenere(ciphertext, key):
    result = ""
    key = key.upper()
    key_index = 0

    for char in ciphertext:
        if char.isalpha():
            k = ord(key[key_index % len(key)]) - 65

            if char.isupper():
                c = ord(char) - 65
                p = (c - k) % 26
                result += chr(p + 65)
            else:
                c = ord(char) - 97
                p = (c - k) % 26
                result += chr(p + 97)

            key_index += 1
        else:
            result += char

    return result
```

```python
#مثال تشغيل
text = "HELLO WORLD"
key = "KEY"

cipher = encrypt_vigenere(text, key)
print("Encrypted:", cipher)

plain = decrypt_vigenere(cipher, key)
print("Decrypted:", plain)
```
