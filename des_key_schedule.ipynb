```python
# دالة الإزاحة الدائرية لليسار
def left_shift(bits, n):
    return bits[n:] + bits[:n]

# توليد المفاتيح الفرعية
def generate_subkeys_des(C0, D0):
    """
    C0, D0: كل منهما 28 بت على شكل سلسلة "010101..."
    ترجع قائمة المفاتيح الفرعية الـ 16
    """
    shifts = [1, 1, 2, 2, 2, 2, 2, 2,
              1, 2, 2, 2, 2, 2, 2, 1]

    C = C0
    D = D0
    subkeys = []

    for i in range(16):
        # تطبيق الإزاحة الدائرية
        C = left_shift(C, shifts[i])
        D = left_shift(D, shifts[i])

        # دمج C و D لإنتاج المفتاح الفرعي (56 بت)
        Ki = C + D
        subkeys.append(Ki)

    return subkeys


```

```python
#مثال تشغيل 
# مفتاح مبدئي وهمي (56 بت)
C0 = "000100110011010001010111011110011001101110111100"
D0 = "111000111010101011110011011001100111000111010101"

subkeys = generate_subkeys_des(C0, D0)

for i, key in enumerate(subkeys, 1):
    print(f"Subkey {i}: {key}")
```
