```python
#التشفير (Encryption)
def encrypt_autokey(plaintext, key):
    key = key.upper()
    plaintext = plaintext.upper()

    result = ""
    key_stream = key + plaintext  # المفتاح الذاتي

    for i, char in enumerate(plaintext):
        if char.isalpha():
            p = ord(char) - 65
            k = ord(key_stream[i]) - 65
            c = (p + k) % 26
            result += chr(c + 65)
        else:
            result += char

    return result

```

```python
#فك التشفير (Decryption)
def decrypt_autokey(ciphertext, key):
    key = key.upper()
    ciphertext = ciphertext.upper()

    result = ""
    key_stream = key

    for i, char in enumerate(ciphertext):
        if char.isalpha():
            c = ord(char) - 65
            k = ord(key_stream[i]) - 65
            p = (c - k) % 26
            plain_char = chr(p + 65)
            result += plain_char
            key_stream += plain_char  # إضافة الحرف المفكوك للمفتاح
        else:
            result += char

    return result
```

```python
#مثال تشغيل
text = "HELLO"
key = "K"

cipher = encrypt_autokey(text, key)
print("Encrypted:", cipher)

plain = decrypt_autokey(cipher, key)
print("Decrypted:", plain)
```
