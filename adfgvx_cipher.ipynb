```python
#تشكيل مصفوفة المفتاح الأول (Polybius Square)
def generate_polybius_square(key):
    key = key.upper()
    key = "".join(c for c in key if c.isalnum())

    seen = set()
    square_string = ""

    for char in key:
        if char not in seen:
            seen.add(char)
            square_string += char

    for char in "ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789":
        if char not in seen:
            square_string += char

    square = [list(square_string[i:i+6]) for i in range(0, 36, 6)]
    return square
```

```python
#مثال تشغيل كود تشكيل مصفوفة المفتاح الاول
square = generate_polybius_square("SECRET")
for row in square:
    print(row)
```

```python
#التشفير (Encryption)


#المرحلة 1: الاستبدال (Substitution)
def adfgvx_substitution(plaintext, square):
    labels = ['A', 'D', 'F', 'G', 'V', 'X']
    plaintext = plaintext.upper()
    plaintext = "".join(c for c in plaintext if c.isalnum())

    result = ""

    for char in plaintext:
        for row in range(6):
            if char in square[row]:
                col = square[row].index(char)
                result += labels[row] + labels[col]
                break

    return result

#المرحلة 2: التبديل العمودي (Columnar Transposition)
def columnar_transposition(text, key):
    key = key.upper()
    cols = len(key)
    rows = (len(text) + cols - 1) // cols

    matrix = [['' for _ in range(cols)] for _ in range(rows)]
    index = 0

    for r in range(rows):
        for c in range(cols):
            if index < len(text):
                matrix[r][c] = text[index]
                index += 1

    sorted_key = sorted(list(enumerate(key)), key=lambda x: x[1])

    cipher = ""
    for idx, _ in sorted_key:
        for r in range(rows):
            if matrix[r][idx] != '':
                cipher += matrix[r][idx]

    return cipher


#التشفير الكامل ADFGVX
def encrypt_adfgvx(plaintext, key1, key2):
    square = generate_polybius_square(key1)
    substituted = adfgvx_substitution(plaintext, square)
    cipher = columnar_transposition(substituted, key2)
    return cipher

```

```python
#مثال تشغيل كامل
plaintext = "HELLO123"
key1 = "SECRET"
key2 = "KEY"

ciphertext = encrypt_adfgvx(plaintext, key1, key2)
print("Encrypted:", ciphertext)
```
