```python
#معالجة النص الأصلي (Plaintext Processing)
def process_plaintext(text):
    text = text.upper()
    text = "".join([c for c in text if c.isalpha()])
    text = text.replace("J", "I")

    processed = ""
    i = 0

    while i < len(text):
        a = text[i]

        if i + 1 < len(text):
            b = text[i + 1]
            if a == b:
                processed += a + "X"
                i += 1
            else:
                processed += a + b
                i += 2
        else:
            processed += a + "X"
            i += 1

    return processed
```

```python
#مثال تشغيل كود المعالجة 
print(process_plaintext("HELLO WORLD"))
```

```python
#تشكيل مصفوفة المفتاح (Key Matrix)
def generate_key_matrix(key):
    key = key.upper()
    key = "".join([c for c in key if c.isalpha()])
    key = key.replace("J", "I")

    seen = set()
    matrix_string = ""

    for char in key:
        if char not in seen:
            seen.add(char)
            matrix_string += char

    for char in "ABCDEFGHIKLMNOPQRSTUVWXYZ":  # بدون J
        if char not in seen:
            matrix_string += char

    matrix = [list(matrix_string[i:i+5]) for i in range(0, 25, 5)]
    return matrix
```

```python
#مثال تشغيل كود تشكيل مصفوفة المفتاح
matrix = generate_key_matrix("KEYWORD")
for row in matrix:
    print(row)
```
